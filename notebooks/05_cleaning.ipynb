```python
import pandas as pd

from covid.common import feature, paths
from covid.common.feature import HighMissingRateDropper, ColumnDropper, QValueSelector

train_data = pd.read_csv(paths.INTERIM_TRAIN_DATA_PATH, dtype={feature.ID: str})
train_data.shape
```

```python
X = train_data.drop(columns=[feature.TARGET])
y = train_data[feature.TARGET]
```

```python
ColumnDropper(columns_to_drop=feature.ID).fit_transform(X).info()
```

```python
HighMissingRateDropper(missing_threshold=0.05).fit_transform(X).shape
```

```python
QValueSelector(q_value_threshold=0.05).fit(X, y).get_feature_names_out()
```
