```python
import pandas as pd

from covid import feature
from covid.constants import INTERIM_TRAIN_DATA_PATH

train_data = pd.read_csv(INTERIM_TRAIN_DATA_PATH, dtype={feature.ID: str})
train_data.shape
```

```python
X = train_data.drop(columns=[feature.TARGET])
y = train_data[feature.TARGET]
```

```python
from covid.feature.column_dropper import ColumnDropper

ColumnDropper(columns_to_drop=feature.ID).fit_transform(X).info()
```

```python
from covid.feature.high_missing_rate_dropper import HighMissingRateDropper

HighMissingRateDropper(missing_threshold=0.05).fit_transform(X).shape
```

```python
from covid.feature.q_value_selector import QValueSelector

QValueSelector(q_value_threshold=0.05).fit(X, y).get_feature_names_out()
```
